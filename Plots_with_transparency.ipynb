# This notebook gives an example of how to create plots with transparency
## And save it as a png file

**Loading necessary packages. Please install packages if not present.**

The plotting function `plot_with_transparency` lives in `plotting_functions.py`, which must be in the same directory as this notebook (or on the Python path).

```python
import numpy as np
from plotting_functions import plot_with_transparency
```

**Creating data to plot**

```python
x=np.linspace(-10,10,num=100)

amp=5
d=2

y=amp*np.sin(x)
yu=(amp+d)*np.sin(x)
yd=(amp-d)*np.sin(x)
```

**Making the plot and saving**

```python
plot_with_transparency(x, y, yu, yd, color='cyan', alpha=.2,
                       filename='sinewtransparency.png', dpi=300);
```
